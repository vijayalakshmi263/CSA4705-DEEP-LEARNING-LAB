```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Given values
n = 20
p = 0.1

# Calculate probability of exactly 3 defective components
probability = binom.pmf(3, n, p)

print("Probability of exactly 3 defective components:")
print(round(probability, 4))

# Calculate PMF for 0 to 20 defective components
x = np.arange(0, n + 1)
pmf = binom.pmf(x, n, p)

# Plot the probability mass function
plt.bar(x, pmf)
plt.xlabel("Number of Defective Components")
plt.ylabel("Probability")
plt.title("Binomial Distribution (n=20, p=0.1)")
plt.xticks(x)
plt.grid(axis='y')
plt.show()

```
