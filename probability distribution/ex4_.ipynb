```python
import numpy as np
import matplotlib.pyplot as plt

C = np.array([[3, 1], [1, 3]])

values, vectors = np.linalg.eig(C)

print("Eigenvalues:", values)
print("Eigenvectors:")
print(vectors)

plt.figure()

plt.arrow(0, 0, vectors[0, 0], vectors[1, 0],
          head_width=0.05, length_includes_head=True,
          label="Eigenvector 1")

plt.arrow(0, 0, vectors[0, 1], vectors[1, 1],
          head_width=0.05, length_includes_head=True,
          label="Eigenvector 2")

plt.xlim(-1, 1)
plt.ylim(-1, 1)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Eigenvectors")
plt.grid()
plt.show()

```
