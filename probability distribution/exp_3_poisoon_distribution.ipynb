```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Average number of customers
mean = 4

# Probability of exactly 6 customers
probability = poisson.pmf(6, mean)

print("Probability of exactly 6 customers:")
print(round(probability, 4))

# Customers from 0 to 12
x = np.arange(0, 13)
pmf = poisson.pmf(x, mean)

# Plot the probability mass function
plt.bar(x, pmf)
plt.xlabel("Number of Customers")
plt.ylabel("Probability")
plt.title("Poisson Distribution (Mean = 4)")
plt.xticks(x)
plt.grid(axis='y')
plt.show()

```
