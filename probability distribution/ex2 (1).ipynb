```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Given values
mean = 70
std_dev = 10

# Calculate probability between 60 and 80
probability = norm.cdf(80, mean, std_dev) - norm.cdf(60, mean, std_dev)

print("Probability of scoring between 60 and 80:")
print(round(probability, 4))

# Generate normal distribution curve
x = np.linspace(40, 100, 500)
y = norm.pdf(x, mean, std_dev)

# Plot the normal distribution
plt.plot(x, y, label="Normal Distribution")

# Highlight the interval between 60 and 80
x_fill = np.linspace(60, 80, 200)
y_fill = norm.pdf(x_fill, mean, std_dev)
plt.fill_between(x_fill, y_fill, alpha=0.4, label="Scores between 60 and 80")

plt.xlabel("Examination Marks")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Examination Marks")
plt.legend()
plt.grid(True)
plt.show()
```
