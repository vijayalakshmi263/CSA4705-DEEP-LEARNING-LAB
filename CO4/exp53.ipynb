```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

text = [
    "deep learning is interesting",
    "deep learning is powerful",
    "machine learning is useful",
    "machine learning is interesting",
    "artificial intelligence is powerful"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(text)

vocab_size = len(tokenizer.word_index) + 1

sequences = []

for sentence in text:
    tokens = tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1, len(tokens)):
        sequences.append(tokens[:i+1])

max_length = max(len(seq) for seq in sequences)

sequences = pad_sequences(
    sequences, maxlen=max_length, padding="pre"
)

X = sequences[:, :-1]
y = sequences[:, -1]

model = Sequential([
    Embedding(vocab_size, 16),
    SimpleRNN(32),
    Dense(vocab_size, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(X, y, epochs=200, verbose=0)

seed_text = "deep learning"
token_list = tokenizer.texts_to_sequences([seed_text])[0]
token_list = pad_sequences(
    [token_list], maxlen=max_length-1, padding="pre"
)

prediction = model.predict(token_list, verbose=0)
next_word = tokenizer.index_word[np.argmax(prediction)]

print("Input Text:", seed_text)
print("Next Predicted Word:", next_word)
```
