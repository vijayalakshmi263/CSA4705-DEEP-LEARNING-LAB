```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

prices = np.array([
    100, 102, 105, 108, 110, 112,
    115, 118, 120, 123, 125, 128,
    130, 133, 135, 138, 140, 143,
    145, 148
]).reshape(-1, 1)

scaler = MinMaxScaler()
data = scaler.fit_transform(prices)

X = []
y = []

for i in range(len(data) - 3):
    X.append(data[i:i+3])
    y.append(data[i+3])

X = np.array(X)
y = np.array(y)

split = int(len(X) * 0.8)

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

model = Sequential([
    LSTM(32, input_shape=(3, 1)),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse")

model.fit(X_train, y_train, epochs=100, verbose=0)

predictions = model.predict(X_test)

actual = scaler.inverse_transform(y_test)
predicted = scaler.inverse_transform(predictions)

print("Actual Prices:", actual.flatten().round(2))
print("Predicted Prices:", predicted.flatten().round(2))

print("MAE:", mean_absolute_error(actual, predicted))
print("MSE:", mean_squared_error(actual, predicted))
print("RMSE:", np.sqrt(mean_squared_error(actual, predicted)))
```
