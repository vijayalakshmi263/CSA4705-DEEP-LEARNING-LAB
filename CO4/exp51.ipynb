```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

prices = np.array([
    100, 102, 105, 103, 108, 110, 112, 115,
    117, 120, 118, 122, 125, 128, 130, 135,
    138, 140, 143, 145
]).reshape(-1, 1)

scaler = MinMaxScaler()
prices = scaler.fit_transform(prices)

X = []
y = []

for i in range(len(prices) - 3):
    X.append(prices[i:i+3])
    y.append(prices[i+3])

X = np.array(X)
y = np.array(y)

split = int(len(X) * 0.8)

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

model = Sequential([
    SimpleRNN(32, activation="tanh", input_shape=(3, 1)),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse")

model.fit(X_train, y_train, epochs=100, verbose=0)

predictions = model.predict(X_test)

actual = scaler.inverse_transform(y_test)
predicted = scaler.inverse_transform(predictions)

print("Actual Stock Prices:", actual.flatten().round(2))
print("Predicted Stock Prices:", predicted.flatten().round(2))

print("MSE:", mean_squared_error(actual, predicted))
print("MAE:", mean_absolute_error(actual, predicted))
```
