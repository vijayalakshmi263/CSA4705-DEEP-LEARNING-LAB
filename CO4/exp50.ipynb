```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

text = "deep learning is very interesting"

chars = sorted(set(text))
char_to_num = {c: i for i, c in enumerate(chars)}
num_to_char = {i: c for i, c in enumerate(chars)}

X = []
y = []

for i in range(len(text) - 4):
    X.append([char_to_num[c] for c in text[i:i+4]])
    y.append(char_to_num[text[i+4]])

X = np.array(X)
y = np.array(y)

X = tf.keras.utils.to_categorical(X, num_classes=len(chars))

model = Sequential([
    SimpleRNN(50, input_shape=(4, len(chars))),
    Dense(len(chars), activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(X, y, epochs=200, verbose=0)

seed = "deep"
generated = seed

for i in range(20):
    seq = [char_to_num[c] for c in generated[-4:]]
    seq = tf.keras.utils.to_categorical(
        [seq], num_classes=len(chars)
    )

    prediction = model.predict(seq, verbose=0)
    next_char = num_to_char[np.argmax(prediction)]

    generated += next_char

print("Generated Text:", generated)
```
