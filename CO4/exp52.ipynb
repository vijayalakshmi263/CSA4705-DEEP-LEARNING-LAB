```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from sklearn.preprocessing import MinMaxScaler

consumption = np.array([
    100, 110, 120, 130, 140, 150,
    160, 170, 180, 190, 200, 210,
    220, 230, 240, 250, 260, 270
]).reshape(-1, 1)

scaler = MinMaxScaler()
data = scaler.fit_transform(consumption)

X = []
y = []

for i in range(len(data) - 3):
    X.append(data[i:i+3])
    y.append(data[i+3])

X = np.array(X)
y = np.array(y)

model = Sequential([
    SimpleRNN(32, activation="tanh", input_shape=(3, 1)),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse")

model.fit(X, y, epochs=100, verbose=0)

predictions = model.predict(X)

actual = scaler.inverse_transform(y)
predicted = scaler.inverse_transform(predictions)

print("Actual Consumption:", actual.flatten().round(2))
print("Predicted Consumption:", predicted.flatten().round(2))

next_input = data[-3:].reshape(1, 3, 1)
next_prediction = model.predict(next_input)

print("Next Day Consumption:",
      scaler.inverse_transform(next_prediction)[0][0])
```
