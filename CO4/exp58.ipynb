```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from sklearn.preprocessing import MinMaxScaler

data = np.array([
    [25, 50, 10],
    [27, 52, 12],
    [29, 54, 14],
    [31, 56, 16],
    [33, 58, 18],
    [35, 60, 20],
    [37, 62, 22],
    [39, 64, 24],
    [41, 66, 26],
    [43, 68, 28]
], dtype=float)

scaler = MinMaxScaler()
data = scaler.fit_transform(data)

X = []
y = []

for i in range(len(data) - 3):
    X.append(data[i:i+3])
    y.append(data[i+3])

X = np.array(X)
y = np.array(y)

model = Sequential([
    SimpleRNN(32, return_sequences=True, input_shape=(3, 3)),
    SimpleRNN(16),
    Dense(3)
])

model.compile(optimizer="adam", loss="mse")

model.fit(X, y, epochs=100, verbose=0)

predictions = model.predict(X)

print("Actual Values:")
print(scaler.inverse_transform(y).round(2))

print("\nPredicted Values:")
print(scaler.inverse_transform(predictions).round(2))

print("\nFinal Loss:", model.evaluate(X, y, verbose=0))
```
