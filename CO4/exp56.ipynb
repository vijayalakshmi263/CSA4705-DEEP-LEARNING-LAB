```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense
from sklearn.preprocessing import MinMaxScaler

data = np.array([
    [25, 60, 1010],
    [26, 62, 1012],
    [27, 65, 1011],
    [28, 63, 1013],
    [29, 67, 1014],
    [30, 70, 1012],
    [31, 68, 1015],
    [32, 72, 1016],
    [33, 74, 1014],
    [34, 75, 1017]
], dtype=float)

scaler = MinMaxScaler()
data = scaler.fit_transform(data)

X = []
y = []

for i in range(len(data) - 3):
    X.append(data[i:i+3])
    y.append(data[i+3])

X = np.array(X)
y = np.array(y)

model = Sequential([
    GRU(32, activation="tanh", input_shape=(3, 3)),
    Dense(3)
])

model.compile(optimizer="adam", loss="mse")

model.fit(X, y, epochs=100, verbose=0)

predictions = model.predict(X)

print("Actual Weather Values:")
print(scaler.inverse_transform(y).round(2))

print("\nPredicted Weather Values:")
print(scaler.inverse_transform(predictions).round(2))

print("\nFinal Loss:", model.evaluate(X, y, verbose=0))
```
