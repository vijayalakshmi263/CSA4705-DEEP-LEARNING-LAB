```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, Dense
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.preprocessing import MinMaxScaler

sales = np.array([
    100, 110, 120, 130, 125, 140,
    150, 160, 155, 170, 180, 190,
    185, 200, 210, 220, 215, 230,
    240, 250
]).reshape(-1, 1)

scaler = MinMaxScaler()
data = scaler.fit_transform(sales)

X = []
y = []

for i in range(len(data) - 3):
    X.append(data[i:i+3])
    y.append(data[i+3])

X = np.array(X)
y = np.array(y)

split = int(len(X) * 0.8)

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# RNN Model
rnn = Sequential([
    SimpleRNN(32, activation="tanh", input_shape=(3, 1)),
    Dense(1)
])

rnn.compile(optimizer="adam", loss="mse")

rnn.fit(X_train, y_train, epochs=100, verbose=0)

rnn_pred = rnn.predict(X_test)

# LSTM Model
lstm = Sequential([
    LSTM(32, input_shape=(3, 1)),
    Dense(1)
])

lstm.compile(optimizer="adam", loss="mse")

lstm.fit(X_train, y_train, epochs=100, verbose=0)

lstm_pred = lstm.predict(X_test)

# Convert predictions to original values
actual = scaler.inverse_transform(y_test)
rnn_result = scaler.inverse_transform(rnn_pred)
lstm_result = scaler.inverse_transform(lstm_pred)

print("RNN MSE:", mean_squared_error(actual, rnn_result))
print("RNN MAE:", mean_absolute_error(actual, rnn_result))

print("\nLSTM MSE:", mean_squared_error(actual, lstm_result))
print("LSTM MAE:", mean_absolute_error(actual, lstm_result))

print("\nActual Sales:", actual.flatten().round(2))
print("RNN Predictions:", rnn_result.flatten().round(2))
print("LSTM Predictions:", lstm_result.flatten().round(2))
```
