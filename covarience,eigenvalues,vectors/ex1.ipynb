```python
import numpy as np

# Study hours and examination marks
study_hours = [2, 4, 6, 8, 10]
marks = [35, 45, 60, 75, 85]

# Calculate covariance
covariance = np.cov(study_hours, marks)

# Display result
print("Covariance Matrix:")
print(covariance)

print("Covariance between Study Hours and Marks:")
print(covariance[0, 1])
```
