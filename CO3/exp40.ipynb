```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (40, 40), (160, 160), 200, -1)
cv2.circle(img, (100, 100), 40, 255, -1)

max_pool = cv2.resize(
    img, (100, 100), interpolation=cv2.INTER_NEAREST
)

avg_pool = cv2.resize(
    img, (100, 100), interpolation=cv2.INTER_AREA
)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")

plt.subplot(1, 3, 2)
plt.imshow(max_pool, cmap="gray")
plt.title("Max Pooling")

plt.subplot(1, 3, 3)
plt.imshow(avg_pool, cmap="gray")
plt.title("Average Pooling")

plt.show()
```
