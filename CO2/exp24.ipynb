```python
import numpy as np
from sklearn.linear_model import Perceptron

# Input data: Income and Credit Score
X = np.array([
    [20, 500],
    [30, 600],
    [40, 650],
    [50, 700],
    [60, 750]
])

# Output: 0 = Ineligible, 1 = Eligible
y = np.array([0, 0, 0, 1, 1])

# Create perceptron model
model = Perceptron(max_iter=1000, random_state=42)

# Train model
model.fit(X, y)

# Predict eligibility
prediction = model.predict([[55, 720]])

print("Income: 55")
print("Credit Score: 720")

if prediction[0] == 1:
    print("Loan Status: Eligible")
else:
    print("Loan Status: Ineligible")
```
