```python
import numpy as np

x = np.array([1.0, 0.5])
target = 1.0

W1 = np.array([[0.2, 0.4], [0.3, 0.1]])
b1 = np.array([0.1, 0.2])
W2 = np.array([[0.5], [0.6]])
b2 = np.array([0.1])

lr = 0.5

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Forward propagation
hidden = sigmoid(x @ W1 + b1)
output = sigmoid(hidden @ W2 + b2)

# Backpropagation
delta_out = (output[0] - target) * output[0] * (1 - output[0])
delta_hidden = (W2[:, 0] * delta_out) * hidden * (1 - hidden)

# Update weights
W2 -= lr * np.outer(hidden, [delta_out])
b2 -= lr * delta_out
W1 -= lr * np.outer(x, delta_hidden)
b1 -= lr * delta_hidden

# Calculate updated output
hidden = sigmoid(x @ W1 + b1)
output = sigmoid(hidden @ W2 + b2)

print("Updated Output:", output[0])
print("Updated Weights:")
print(W1)
print(W2)
```
