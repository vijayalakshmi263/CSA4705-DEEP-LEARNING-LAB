```python
import numpy as np

# Traffic states: Low, Medium, High
# Actions: Road A Green, Road B Green

rewards = np.array([
    [10, 2],
    [5, 8],
    [1, 12]
])

# Initialize Q-table
Q = np.zeros((3, 2))

learning_rate = 0.1
gamma = 0.9

# Q-learning
for episode in range(1000):
    for state in range(3):
        for action in range(2):

            reward = rewards[state, action]

            Q[state, action] += learning_rate * (
                reward + gamma * np.max(Q[state]) - Q[state, action]
            )

# Display results
states = ["Low Traffic", "Medium Traffic", "High Traffic"]
actions = ["Road A Green", "Road B Green"]

for i in range(3):
    best_action = np.argmax(Q[i])
    print(states[i], ":", actions[best_action])

print("\nQ-Table:")
print(np.round(Q, 2))

print("\nOptimal traffic signals learned successfully")
```
