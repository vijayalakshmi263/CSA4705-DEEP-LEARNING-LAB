```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

# Generate dataset
X, y = make_classification(
    n_samples=200,
    n_features=20,
    random_state=42
)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Create neural network
model = MLPClassifier(
    hidden_layer_sizes=(100, 100),
    max_iter=2000,
    random_state=42
)

# Train model
model.fit(X_train, y_train)

# Predict results
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

# Calculate errors
train_error = 1 - accuracy_score(y_train, train_pred)
test_error = 1 - accuracy_score(y_test, test_pred)

print("Training Error:", round(train_error, 3))
print("Testing Error:", round(test_error, 3))

if test_error > train_error + 0.15:
    print("Overfitting detected")
else:
    print("No significant overfitting detected")

print("Improvements:")
print("1. Use regularization")
print("2. Apply dropout")
print("3. Use early stopping")
print("4. Increase training data")
```
