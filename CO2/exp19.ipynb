```python
import numpy as np
from sklearn.linear_model import Lasso, Ridge
from sklearn.metrics import mean_squared_error

# Financial dataset
X = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50],
    [40, 50, 60],
    [50, 60, 70]
])

y = np.array([100, 150, 200, 250, 300])

# L1 Regularization
l1 = Lasso(alpha=0.1)
l1.fit(X, y)

# L2 Regularization
l2 = Ridge(alpha=0.1)
l2.fit(X, y)

# Predictions
pred1 = l1.predict(X)
pred2 = l2.predict(X)

print("L1 Coefficients:", np.round(l1.coef_, 2))
print("L2 Coefficients:", np.round(l2.coef_, 2))

print("L1 Error:", round(mean_squared_error(y, pred1), 2))
print("L2 Error:", round(mean_squared_error(y, pred2), 2))
```
