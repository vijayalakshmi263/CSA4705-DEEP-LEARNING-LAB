```python
# Different learning rates
learning_rates = [0.01, 0.1, 0.5]

for lr in learning_rates:

    # Initial parameter
    x = 10

    # Gradient descent
    for i in range(50):
        gradient = 2 * x
        x = x - lr * gradient

    # Calculate final loss
    loss = x ** 2

    print("Learning Rate:", lr)
    print("Final Loss:", round(loss, 6))

    if loss < 0.01:
        print("Convergence is good")
    else:
        print("Convergence is slow or unstable")

    print()
```
