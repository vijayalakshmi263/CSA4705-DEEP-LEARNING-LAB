```python
import numpy as np

# Input data
X = np.array([[1, 0]])

# Weights and biases
W1 = np.array([[0.5, 0.2],
               [0.3, 0.4]])
b1 = np.array([0.1, 0.1])

W2 = np.array([[0.6],
               [0.7]])
b2 = np.array([0.2])

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Hidden layer
hidden = sigmoid(np.dot(X, W1) + b1)

# Output layer
output = sigmoid(np.dot(hidden, W2) + b2)

print("Hidden Layer Output:", np.round(hidden, 3))
print("Final Output:", np.round(output, 3))
```
