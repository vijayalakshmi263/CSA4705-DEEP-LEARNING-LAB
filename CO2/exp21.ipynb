```python
import numpy as np

# Floating-point values
a = np.float32(0.1)
b = np.float32(0.2)

c = np.float64(0.1)
d = np.float64(0.2)

# Addition
result1 = a + b
result2 = c + d

print("Float32 Result:", result1)
print("Float64 Result:", result2)

# Precision comparison
print("Float32 Precision:", np.finfo(np.float32).eps)
print("Float64 Precision:", np.finfo(np.float64).eps)

# Large number calculation
x = np.float32(10000000)
y = np.float32(1)

print("Float32 Large Number Result:", x + y)
```
