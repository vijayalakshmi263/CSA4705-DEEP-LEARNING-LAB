```python
import numpy as np

actual = np.array([1, 0, 1, 0])
predicted = np.array([0.9, 0.2, 0.8, 0.1])

eps = 1e-9

loss = -np.mean(
    actual * np.log(predicted + eps) +
    (1 - actual) * np.log(1 - predicted + eps)
)

labels = (predicted >= 0.5).astype(int)

print("Binary Cross Entropy Loss:", round(loss, 4))
print("Predicted Classes:", labels)
print("Correct Predictions:", np.sum(labels == actual), "out of", len(actual))
```
