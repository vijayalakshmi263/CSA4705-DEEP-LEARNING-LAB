```python
import numpy as np

# Movie feature vectors
A = np.array([2, 4, 6])
B = np.array([1, 3, 5])

# Vector addition
combined = A + B

# Calculate dot product
similarity = np.dot(A, B)

# Display results
print("Combined Feature Vector:")
print(combined)

print("Dot Product (Similarity Score):")
print(similarity)
```
