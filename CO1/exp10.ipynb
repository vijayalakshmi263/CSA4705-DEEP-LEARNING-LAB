```python
import numpy as np
import matplotlib.pyplot as plt

# Generate random temperature samples
np.random.seed(42)
temperature = np.random.normal(30, 5, 1000)

# Calculate mean and standard deviation
mean = np.mean(temperature)
std = np.std(temperature)

print("First 10 Temperature Samples:")
print(np.round(temperature[:10], 2))

print("Mean Temperature:", round(mean, 2))
print("Standard Deviation:", round(std, 2))

# Plot probability distribution
plt.hist(temperature, bins=20, density=True)
plt.xlabel("Temperature")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Temperature")
plt.show()

print("Observations approximately follow normal distribution")
```
