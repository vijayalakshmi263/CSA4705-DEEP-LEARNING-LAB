```python
import numpy as np

# Define transition matrix
A = np.array([[0.5, 0.2, 0.3],
              [0.3, 0.5, 0.2],
              [0.2, 0.3, 0.5]])

# Calculate eigenvalues and eigenvectors
values, vectors = np.linalg.eig(A)

print("Transition Matrix:")
print(A)

print("Eigenvalues:")
print(np.round(values, 2))

print("Eigenvectors:")
print(np.round(vectors, 2))

# Find the most important web page
index = np.argmax(values.real)
importance = np.abs(vectors[:, index].real)

print("Page Importance:")
print(np.round(importance / np.sum(importance), 2))

print("Most Important Page: Page", np.argmax(importance) + 1)
```
