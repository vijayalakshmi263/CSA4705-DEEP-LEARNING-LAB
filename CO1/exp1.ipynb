```python
import numpy as np

# Initial position of robot
A = np.array([2, 3, 4])

# Movement vector
B = np.array([1, 2, 3])

# Scalar value
k = 2

# Vector addition
addition = A + B

# Vector subtraction
subtraction = A - B

# Scalar multiplication
multiplication = k * A

print("Initial Position:", A)
print("Movement Vector:", B)
print("Vector Addition:", addition)
print("Vector Subtraction:", subtraction)
print("Scalar Multiplication:", multiplication)
```
