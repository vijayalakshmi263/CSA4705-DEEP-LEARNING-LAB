```python
# Given probabilities
prevalence = 0.01
sensitivity = 0.95
specificity = 0.90

# Calculate false positive rate
false_positive = 1 - specificity

# Apply Bayes' theorem
positive_probability = (prevalence * sensitivity) + ((1 - prevalence) * false_positive)

result = (prevalence * sensitivity) / positive_probability

print("Disease Prevalence:", prevalence)
print("Sensitivity:", sensitivity)
print("Specificity:", specificity)

print("Probability of having disease after positive test:",
      round(result, 4))

print("Percentage:", round(result * 100, 2), "%")
```
