```python
import numpy as np

# Define two matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Matrix multiplication
C = np.dot(A, B)

# Transpose of matrix
T = A.T

print("Matrix A:")
print(A)

print("Matrix B:")
print(B)

print("Matrix Multiplication:")
print(C)

print("Transpose of Matrix A:")
print(T)
```
