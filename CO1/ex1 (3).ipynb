```python
import numpy as np

# Initial robot position
A = np.array([1, 2, 3])

# Robot movement vector
B = np.array([4, 5, 6])

# Vector Addition
addition = A + B

# Vector Subtraction
subtraction = A - B

# Scalar Multiplication
scalar = 2 * A

# New Robot Position
new_position = A + B

print("Vector A:", A)
print("Vector B:", B)
print("Vector Addition:", addition)
print("Vector Subtraction:", subtraction)
print("Scalar Multiplication:", scalar)
print("New Robot Position:", new_position)
```
