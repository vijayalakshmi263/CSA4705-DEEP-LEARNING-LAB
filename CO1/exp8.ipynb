```python
# Probability of defective products
P_A = 0.6
P_B = 0.4

# Defect rates
P_D_A = 0.02
P_D_B = 0.05

# Total probability of defective product
P_D = (P_A * P_D_A) + (P_B * P_D_B)

# Bayes theorem
P_A_D = (P_A * P_D_A) / P_D
P_B_D = (P_B * P_D_B) / P_D

print("Probability of defective product:", round(P_D, 3))
print("Probability from Machine A:", round(P_A_D, 3))
print("Probability from Machine B:", round(P_B_D, 3))
```
