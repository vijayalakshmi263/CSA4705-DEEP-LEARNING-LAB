```python
# Probability of products manufactured by machines
P_A = 0.6
P_B = 0.4

# Defect rates
P_D_A = 0.02
P_D_B = 0.05

# Total probability of defective products
P_D = (P_A * P_D_A) + (P_B * P_D_B)

# Conditional probability
P_A_D = (P_A * P_D_A) / P_D

print("Probability of Machine A:", P_A)
print("Probability of Machine B:", P_B)

print("Probability of Defective Product:", round(P_D, 3))

print("Probability that defective product came from Machine A:",
      round(P_A_D, 3))
```
