```python
import numpy as np

# Customer purchasing data
A = np.array([2, 3])
B = np.array([5, 7])

# Euclidean distance
euclidean = np.linalg.norm(A - B)

# Manhattan distance
manhattan = np.sum(np.abs(A - B))

print("Customer A:", A)
print("Customer B:", B)

print("Euclidean Distance:", round(euclidean, 2))
print("Manhattan Distance:", manhattan)

if euclidean < manhattan:
    print("Euclidean distance is smaller")
else:
    print("Manhattan distance is smaller")
```
