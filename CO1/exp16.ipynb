```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# House features: Area, Bedrooms, Location
X = np.array([
    [1000, 2, 1],
    [1500, 3, 2],
    [2000, 4, 2],
    [2500, 4, 3],
    [3000, 5, 3]
])

# House prices (in lakhs)
y = np.array([30, 45, 60, 75, 90])

# Create and train model
model = LinearRegression()
model.fit(X, y)

# Predict house price
prediction = model.predict([[1800, 3, 2]])

# Calculate prediction error
predicted = model.predict(X)
error = mean_absolute_error(y, predicted)

print("Predicted House Price:", round(prediction[0], 2), "Lakhs")
print("Mean Absolute Error:", round(error, 2))
```
