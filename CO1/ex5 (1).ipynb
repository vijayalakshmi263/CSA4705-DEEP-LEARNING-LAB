```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Customer preference vectors
A = np.array([1, 2, 3])
B = np.array([2, 3, 4])
C = np.array([4, 1, 2])

# Dot Product
dot_AB = np.dot(A, B)
dot_AC = np.dot(A, C)

# Cosine Similarity
cos_AB = cosine_similarity([A], [B])
cos_AC = cosine_similarity([A], [C])

# Display Results
print("Customer A:", A)
print("Customer B:", B)
print("Customer C:", C)

print("\nDot Product:")
print("A and B:", dot_AB)
print("A and C:", dot_AC)

print("\nCosine Similarity:")
print("A and B:", round(cos_AB[0][0], 2))
print("A and C:", round(cos_AC[0][0], 2))

if cos_AB[0][0] > cos_AC[0][0]:
    print("\nCustomer A and B have similar interests.")
else:
    print("\nCustomer A and C have similar interests.")
```
