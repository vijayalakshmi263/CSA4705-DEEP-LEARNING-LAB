```python
import numpy as np

# Customer preference vectors
A = np.array([2, 3, 4])
B = np.array([3, 4, 5])

# Calculate dot product
dot = np.dot(A, B)

# Calculate cosine similarity
cosine = dot / (np.linalg.norm(A) * np.linalg.norm(B))

print("Customer A:", A)
print("Customer B:", B)

print("Dot Product:", dot)
print("Cosine Similarity:", round(cosine, 2))

if cosine > 0.9:
    print("Customers have similar interests")
else:
    print("Customers have different interests")
```
