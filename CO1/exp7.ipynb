```python
import numpy as np

# Student examination marks
marks = np.array([70, 80, 90, 85, 75])

# Calculate mean
mean = np.mean(marks)

# Calculate variance
variance = np.var(marks)

# Calculate standard deviation
std = np.std(marks)

print("Student Marks:", marks)
print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", round(std, 2))

if std < 10:
    print("Student performance is consistent")
else:
    print("Student performance is not consistent")
```
