```python
import numpy as np

# Define two matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Matrix Multiplication
multiplication = np.dot(A, B)

# Matrix Transpose
transpose = A.T

# Display results
print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nMatrix Multiplication:")
print(multiplication)

print("\nTranspose of Matrix A:")
print(transpose)
```
