```python
# Initial parameter
x = 10

# Learning rate
lr = 0.1

# Gradient descent iterations
for i in range(20):

    # Gradient of loss function
    gradient = 2 * x

    # Update parameter
    x = x - lr * gradient

# Calculate minimum cost
cost = x ** 2

print("Optimal Value:", round(x, 4))
print("Minimum Cost:", round(cost, 4))
```
