```python
import numpy as np

# Define encryption matrix
A = np.array([[2, 3],
              [1, 4]])

# Calculate determinant
det = np.linalg.det(A)

# Calculate inverse
inv = np.linalg.inv(A)

# Calculate rank
rank = np.linalg.matrix_rank(A)

print("Encryption Matrix:")
print(A)

print("Determinant:", round(det, 2))

print("Inverse Matrix:")
print(np.round(inv, 2))

print("Rank:", rank)

if det != 0:
    print("Matrix is suitable for encryption")
else:
    print("Matrix is not suitable for encryption")
```
